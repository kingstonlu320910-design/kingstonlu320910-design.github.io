```python
# ============================================================
# CAT PROBLEMATIC-BEHAVIOR ANALYSIS
# Google Colab Version
# ============================================================

# Install required packages if needed
!pip -q install openpyxl statsmodels

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import glob
import json
import shutil
import warnings
import zipfile

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import chi2_contingency, spearmanr, kruskal

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, label_binarize
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    log_loss,
    roc_auc_score,
    roc_curve,
    auc
)

import statsmodels.api as sm

warnings.filterwarnings("ignore")


# ============================================================
# SETTINGS
# ============================================================

TARGET_COLUMN = "behavior_Issue"
SEX_COLUMN = "SEX"
AGE_COLUMN = "Age"
GROUP_COLUMN = "In_A_Group"

BEHAVIOR_ORDER = [
    "No",
    "A_little",
    "Some",
    "A_lot"
]


# ============================================================
# 3. LOCATE AND LOAD THE DATA
# ============================================================

# Find CSV files in /content
csv_files = glob.glob("/content/*.csv")

if len(csv_files) == 0:
    raise FileNotFoundError(
        "No CSV file was found in /content. "
        "Please upload the CSV file to Google Colab."
    )

print("CSV files found:")
for i, file in enumerate(csv_files):
    print(f"{i}: {file}")

# Automatically use the first CSV file.
# If several CSV files exist, replace csv_files[0] with the correct one.
FILE_PATH = csv_files[0]

print("\nUsing file:")
print(FILE_PATH)

# Try common encodings
try:
    df = pd.read_csv(FILE_PATH)
except UnicodeDecodeError:
    df = pd.read_csv(FILE_PATH, encoding="latin-1")

print("\nOriginal data shape:", df.shape)
print("\nOriginal columns:")
print(df.columns.tolist())


# ============================================================
# 4. CLEAN COLUMN NAMES
# ============================================================

# Remove leading/trailing spaces
df.columns = df.columns.str.strip()

# Replace spaces inside column names with underscores
df.columns = [
    str(column).strip().replace(" ", "_")
    for column in df.columns
]

print("\nCleaned columns:")
print(df.columns.tolist())


# ============================================================
# 5. CHECK REQUIRED COLUMNS
# ============================================================

required_columns = [
    TARGET_COLUMN,
    SEX_COLUMN,
    AGE_COLUMN,
    GROUP_COLUMN
]

missing_columns = [
    column for column in required_columns
    if column not in df.columns
]

if missing_columns:
    raise ValueError(
        f"The following required columns are missing: {missing_columns}\n"
        f"Available columns are: {df.columns.tolist()}"
    )


# ============================================================
# 6. BASIC DATA CLEANING
# ============================================================

# Convert blank strings and common missing-value words to NaN
missing_words = [
    "",
    " ",
    "unknown",
    "Unknown",
    "UNKNOWN",
    "na",
    "NA",
    "N/A",
    "n/a",
    "none",
    "None",
    "NULL",
    "null"
]

df = df.replace(missing_words, np.nan)

# Clean categorical columns
for column in [TARGET_COLUMN, SEX_COLUMN, GROUP_COLUMN]:
    df[column] = df[column].astype("string").str.strip()

# Convert age to numeric
# Values such as "unknown" become NaN
df[AGE_COLUMN] = pd.to_numeric(
    df[AGE_COLUMN],
    errors="coerce"
)

# Remove rows with missing target values
df = df.dropna(subset=[TARGET_COLUMN]).copy()

# Keep only observed behavior categories
observed_behavior_levels = [
    level for level in BEHAVIOR_ORDER
    if level in df[TARGET_COLUMN].dropna().unique()
]

# Add any unexpected levels to the end
unexpected_levels = [
    level
    for level in df[TARGET_COLUMN].dropna().unique()
    if level not in observed_behavior_levels
]

behavior_levels = observed_behavior_levels + unexpected_levels

df[TARGET_COLUMN] = pd.Categorical(
    df[TARGET_COLUMN],
    categories=behavior_levels,
    ordered=True
)

print("\nBehavior levels used:")
print(behavior_levels)

print("\nCleaned data shape:", df.shape)

# Save cleaned data
df.to_csv(
    os.path.join(OUTPUT_FOLDER, "cleaned_data.csv"),
    index=False
)


# ============================================================
# 7. DESCRIPTIVE INFORMATION
# ============================================================

data_summary = pd.DataFrame({
    "Variable": df.columns,
    "Data_Type": [str(df[column].dtype) for column in df.columns],
    "Missing_Count": [df[column].isna().sum() for column in df.columns],
    "Missing_Percent": [
        100 * df[column].isna().mean()
        for column in df.columns
    ],
    "Unique_Values": [
        df[column].nunique(dropna=True)
        for column in df.columns
    ]
})

data_summary.to_csv(
    os.path.join(OUTPUT_FOLDER, "data_summary.csv"),
    index=False
)

behavior_counts = (
    df[TARGET_COLUMN]
    .value_counts(dropna=False)
    .reindex(behavior_levels)
    .rename_axis("Behavior_Level")
    .reset_index(name="Count")
)

behavior_counts["Percent"] = (
    100 * behavior_counts["Count"] / behavior_counts["Count"].sum()
)

behavior_counts.to_csv(
    os.path.join(OUTPUT_FOLDER, "behavior_level_counts.csv"),
    index=False
)


# Behavior-frequency plot
plt.figure(figsize=(8, 5))
plt.bar(
    behavior_counts["Behavior_Level"].astype(str),
    behavior_counts["Count"]
)
plt.title("Distribution of Problematic Behavior")
plt.xlabel("Problematic Behavior Level")
plt.ylabel("Number of Cats")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_behavior_distribution.png"
    ),
    dpi=300,
    bbox_inches="tight"
)
plt.close()


# ============================================================
# 8. CRAMÉR'S V FUNCTION
# ============================================================

def cramers_v_bias_corrected(confusion_table):
    """
    Calculate bias-corrected Cramér's V.

    Cramér's V ranges from 0 to 1:
    0 = little or no association
    1 = very strong association
    """

    chi2 = chi2_contingency(confusion_table)[0]
    n = confusion_table.to_numpy().sum()

    if n <= 1:
        return np.nan

    phi2 = chi2 / n

    rows, columns = confusion_table.shape

    phi2_corrected = max(
        0,
        phi2 - ((columns - 1) * (rows - 1)) / (n - 1)
    )

    rows_corrected = (
        rows - ((rows - 1) ** 2) / (n - 1)
    )

    columns_corrected = (
        columns - ((columns - 1) ** 2) / (n - 1)
    )

    denominator = min(
        columns_corrected - 1,
        rows_corrected - 1
    )

    if denominator <= 0:
        return np.nan

    return np.sqrt(phi2_corrected / denominator)


# ============================================================
# 9. SEX VS. PROBLEMATIC BEHAVIOR
# ============================================================

sex_data = df[
    [SEX_COLUMN, TARGET_COLUMN]
].dropna().copy()

sex_table = pd.crosstab(
    sex_data[SEX_COLUMN],
    sex_data[TARGET_COLUMN],
    dropna=False
)

sex_table.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "sex_vs_behavior_contingency_table.csv"
    )
)

sex_chi2, sex_p, sex_dof, sex_expected = chi2_contingency(
    sex_table
)

sex_cramers_v = cramers_v_bias_corrected(sex_table)

sex_expected_df = pd.DataFrame(
    sex_expected,
    index=sex_table.index,
    columns=sex_table.columns
)

sex_expected_df.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "sex_vs_behavior_expected_counts.csv"
    )
)

sex_results = pd.DataFrame({
    "Analysis": ["SEX vs problematic_behavior"],
    "Sample_Size": [len(sex_data)],
    "Chi_Square": [sex_chi2],
    "Degrees_of_Freedom": [sex_dof],
    "P_Value": [sex_p],
    "Cramers_V": [sex_cramers_v]
})

sex_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "sex_vs_behavior_statistics.csv"
    ),
    index=False
)

# Row percentages
sex_percent = sex_table.div(
    sex_table.sum(axis=1),
    axis=0
) * 100

sex_percent.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "sex_vs_behavior_row_percentages.csv"
    )
)

# Stacked percentage plot
sex_percent.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 6)
)

plt.title("Problematic Behavior Distribution by Sex")
plt.xlabel("Sex")
plt.ylabel("Percentage Within Sex")
plt.legend(
    title="Problematic Behavior",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_sex_vs_behavior_stacked_bar.png"
    ),
    dpi=300,
    bbox_inches="tight"
)
plt.close()


# ============================================================
# 10. IN-A-GROUP VS. PROBLEMATIC BEHAVIOR
# ============================================================

group_data = df[
    [GROUP_COLUMN, TARGET_COLUMN]
].dropna().copy()

group_table = pd.crosstab(
    group_data[GROUP_COLUMN],
    group_data[TARGET_COLUMN],
    dropna=False
)

group_table.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "group_vs_behavior_contingency_table.csv"
    )
)

group_chi2, group_p, group_dof, group_expected = chi2_contingency(
    group_table
)

group_cramers_v = cramers_v_bias_corrected(group_table)

group_expected_df = pd.DataFrame(
    group_expected,
    index=group_table.index,
    columns=group_table.columns
)

group_expected_df.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "group_vs_behavior_expected_counts.csv"
    )
)

group_results = pd.DataFrame({
    "Analysis": ["In_A_Group vs problematic_behavior"],
    "Sample_Size": [len(group_data)],
    "Chi_Square": [group_chi2],
    "Degrees_of_Freedom": [group_dof],
    "P_Value": [group_p],
    "Cramers_V": [group_cramers_v]
})

group_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "group_vs_behavior_statistics.csv"
    ),
    index=False
)

group_percent = group_table.div(
    group_table.sum(axis=1),
    axis=0
) * 100

group_percent.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "group_vs_behavior_row_percentages.csv"
    )
)

# Stacked percentage plot
group_percent.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 6)
)

plt.title("Problematic Behavior Distribution by Group Status")
plt.xlabel("In a Group")
plt.ylabel("Percentage Within Group Status")
plt.legend(
    title="Problematic Behavior",
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_group_vs_behavior_stacked_bar.png"
    ),
    dpi=300,
    bbox_inches="tight"
)
plt.close()


# ============================================================
# 11. AGE VS. PROBLEMATIC BEHAVIOR
# ============================================================

age_data = df[
    [AGE_COLUMN, TARGET_COLUMN]
].dropna().copy()

# Convert ordered behavior categories to numerical ranks
behavior_rank_map = {
    level: rank
    for rank, level in enumerate(behavior_levels)
}

age_data["behavior_rank"] = (
    age_data[TARGET_COLUMN]
    .astype(str)
    .map(behavior_rank_map)
)

# Spearman rank correlation
age_spearman_rho, age_spearman_p = spearmanr(
    age_data[AGE_COLUMN],
    age_data["behavior_rank"],
    nan_policy="omit"
)

# Kruskal-Wallis test:
# Tests whether age distributions differ across behavior groups.
age_groups = []

valid_age_group_names = []

for level in behavior_levels:
    values = age_data.loc[
        age_data[TARGET_COLUMN].astype(str) == level,
        AGE_COLUMN
    ].dropna()

    if len(values) > 0:
        age_groups.append(values)
        valid_age_group_names.append(level)

if len(age_groups) >= 2:
    age_kruskal_h, age_kruskal_p = kruskal(*age_groups)
else:
    age_kruskal_h = np.nan
    age_kruskal_p = np.nan

age_results = pd.DataFrame({
    "Analysis": ["Age vs problematic_behavior"],
    "Sample_Size": [len(age_data)],
    "Spearman_Rho": [age_spearman_rho],
    "Spearman_P_Value": [age_spearman_p],
    "Kruskal_Wallis_H": [age_kruskal_h],
    "Kruskal_Wallis_P_Value": [age_kruskal_p]
})

age_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "age_vs_behavior_statistics.csv"
    ),
    index=False
)

# Age descriptive statistics by behavior category
age_descriptive = (
    age_data.groupby(
        TARGET_COLUMN,
        observed=False
    )[AGE_COLUMN]
    .agg([
        "count",
        "mean",
        "std",
        "median",
        "min",
        "max"
    ])
    .reset_index()
)

age_descriptive.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "age_by_behavior_descriptive_statistics.csv"
    ),
    index=False
)

# Age boxplot
boxplot_data = [
    age_data.loc[
        age_data[TARGET_COLUMN].astype(str) == level,
        AGE_COLUMN
    ].dropna()
    for level in behavior_levels
]

plt.figure(figsize=(9, 6))
plt.boxplot(
    boxplot_data,
    tick_labels=behavior_levels,
    showmeans=True
)
plt.title("Age by Problematic Behavior Level")
plt.xlabel("Problematic Behavior Level")
plt.ylabel("Age")
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_age_vs_behavior_boxplot.png"
    ),
    dpi=300,
    bbox_inches="tight"
)
plt.close()

# Age jitter-style scatter plot
plt.figure(figsize=(9, 6))

rng = np.random.default_rng(42)

for category_number, level in enumerate(behavior_levels):
    current_ages = age_data.loc[
        age_data[TARGET_COLUMN].astype(str) == level,
        AGE_COLUMN
    ].dropna()

    jittered_x = (
        np.repeat(category_number, len(current_ages))
        + rng.normal(0, 0.06, len(current_ages))
    )

    plt.scatter(
        jittered_x,
        current_ages,
        alpha=0.6
    )

plt.xticks(
    range(len(behavior_levels)),
    behavior_levels,
    rotation=30
)
plt.title("Individual Ages by Problematic Behavior Level")
plt.xlabel("Problematic Behavior Level")
plt.ylabel("Age")
plt.tight_layout()
plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_age_vs_behavior_scatter.png"
    ),
    dpi=300,
    bbox_inches="tight"
)
plt.close()


# ============================================================
# 12. COMBINE INDIVIDUAL ASSOCIATION RESULTS
# ============================================================

combined_association_results = pd.DataFrame([
    {
        "Predictor": SEX_COLUMN,
        "Predictor_Type": "Categorical",
        "Primary_Test": "Chi-square",
        "Statistic": sex_chi2,
        "P_Value": sex_p,
        "Effect_Size_Name": "Bias-corrected Cramer's V",
        "Effect_Size": sex_cramers_v,
        "Sample_Size": len(sex_data)
    },
    {
        "Predictor": AGE_COLUMN,
        "Predictor_Type": "Continuous",
        "Primary_Test": "Spearman rank correlation",
        "Statistic": age_spearman_rho,
        "P_Value": age_spearman_p,
        "Effect_Size_Name": "Spearman rho",
        "Effect_Size": age_spearman_rho,
        "Sample_Size": len(age_data)
    },
    {
        "Predictor": GROUP_COLUMN,
        "Predictor_Type": "Categorical",
        "Primary_Test": "Chi-square",
        "Statistic": group_chi2,
        "P_Value": group_p,
        "Effect_Size_Name": "Bias-corrected Cramer's V",
        "Effect_Size": group_cramers_v,
        "Sample_Size": len(group_data)
    }
])

combined_association_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "all_individual_association_results.csv"
    ),
    index=False
)


# ============================================================
# 13. PREPARE MULTINOMIAL LOGISTIC REGRESSION
# ============================================================

# Use every column except the target as a predictor.
predictor_columns = [
    column
    for column in df.columns
    if column != TARGET_COLUMN
]

X = df[predictor_columns].copy()
y = df[TARGET_COLUMN].astype(str).copy()

# Remove columns containing only missing values
all_missing_columns = [
    column
    for column in X.columns
    if X[column].isna().all()
]

if all_missing_columns:
    print("\nRemoving all-missing predictor columns:")
    print(all_missing_columns)
    X = X.drop(columns=all_missing_columns)

# Convert columns that are mostly numeric into numeric columns.
# This helps when numeric values were read as text.
for column in X.columns:
    if X[column].dtype == "object" or str(X[column].dtype).startswith("string"):
        converted = pd.to_numeric(X[column], errors="coerce")

        original_nonmissing = X[column].notna().sum()
        converted_nonmissing = converted.notna().sum()

        if (
            original_nonmissing > 0
            and converted_nonmissing / original_nonmissing >= 0.90
        ):
            X[column] = converted

# Identify numerical and categorical predictors
numerical_predictors = X.select_dtypes(
    include=[np.number]
).columns.tolist()

categorical_predictors = [
    column
    for column in X.columns
    if column not in numerical_predictors
]

print("\nNumerical predictors:")
print(numerical_predictors)

print("\nCategorical predictors:")
print(categorical_predictors)

predictor_types = pd.DataFrame({
    "Predictor": X.columns,
    "Type": [
        "Numerical"
        if column in numerical_predictors
        else "Categorical"
        for column in X.columns
    ]
})

predictor_types.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_predictor_types.csv"
    ),
    index=False
)


# ============================================================
# 14. CREATE PREPROCESSING PIPELINE
# ============================================================

numeric_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                drop="first",
                sparse_output=False
            )
        )
    ]
)

transformers = []

if numerical_predictors:
    transformers.append(
        (
            "numeric",
            numeric_pipeline,
            numerical_predictors
        )
    )

if categorical_predictors:
    transformers.append(
        (
            "categorical",
            categorical_pipeline,
            categorical_predictors
        )
    )

preprocessor = ColumnTransformer(
    transformers=transformers,
    remainder="drop"
)


# ============================================================
# 15. TRAIN-TEST SPLIT
# ============================================================

# Stratification can fail when a class has very few observations.
class_counts = y.value_counts()

can_stratify = (
    class_counts.min() >= 2
    and len(df) >= 20
)

stratify_variable = y if can_stratify else None

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=stratify_variable
)

print("\nTraining sample size:", len(X_train))
print("Testing sample size:", len(X_test))


# ============================================================
# 16. FIT MULTINOMIAL LOGISTIC REGRESSION
# ============================================================

multinomial_model = Pipeline(
    steps=[
        (
            "preprocessor",
            preprocessor
        ),
        (
            "model",
            LogisticRegression(
                solver="lbfgs",
                max_iter=5000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

multinomial_model.fit(X_train, y_train)


# ============================================================
# 17. MODEL PREDICTIONS AND PERFORMANCE
# ============================================================

y_train_pred = multinomial_model.predict(X_train)
y_test_pred = multinomial_model.predict(X_test)

y_train_probability = multinomial_model.predict_proba(X_train)
y_test_probability = multinomial_model.predict_proba(X_test)

model_classes = multinomial_model.named_steps[
    "model"
].classes_

training_accuracy = accuracy_score(
    y_train,
    y_train_pred
)

testing_accuracy = accuracy_score(
    y_test,
    y_test_pred
)

training_balanced_accuracy = balanced_accuracy_score(
    y_train,
    y_train_pred
)

testing_balanced_accuracy = balanced_accuracy_score(
    y_test,
    y_test_pred
)

training_log_loss = log_loss(
    y_train,
    y_train_probability,
    labels=model_classes
)

testing_log_loss = log_loss(
    y_test,
    y_test_probability,
    labels=model_classes
)

performance_results = pd.DataFrame({
    "Metric": [
        "Training accuracy",
        "Testing accuracy",
        "Training balanced accuracy",
        "Testing balanced accuracy",
        "Training log loss",
        "Testing log loss"
    ],
    "Value": [
        training_accuracy,
        testing_accuracy,
        training_balanced_accuracy,
        testing_balanced_accuracy,
        training_log_loss,
        testing_log_loss
    ]
})

performance_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_model_performance.csv"
    ),
    index=False
)


# Classification reports
train_report = classification_report(
    y_train,
    y_train_pred,
    output_dict=True,
    zero_division=0
)

test_report = classification_report(
    y_test,
    y_test_pred,
    output_dict=True,
    zero_division=0
)

pd.DataFrame(train_report).transpose().to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_training_classification_report.csv"
    )
)

pd.DataFrame(test_report).transpose().to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_testing_classification_report.csv"
    )
)


# Save individual test predictions
test_predictions = X_test.copy()

test_predictions["Actual_Behavior"] = y_test.values
test_predictions["Predicted_Behavior"] = y_test_pred

for class_index, class_name in enumerate(model_classes):
    test_predictions[
        f"Probability_{class_name}"
    ] = y_test_probability[:, class_index]

test_predictions.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_test_predictions.csv"
    ),
    index=False
)


# ============================================================
# 18. CONFUSION MATRICES
# ============================================================

# Raw-count confusion matrix
fig, ax = plt.subplots(figsize=(8, 7))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_test_pred,
    labels=model_classes,
    xticks_rotation=30,
    cmap=None,
    ax=ax
)

ax.set_title("Multinomial Logistic Regression: Confusion Matrix")
plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_multinomial_confusion_matrix_counts.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# Normalized confusion matrix
fig, ax = plt.subplots(figsize=(8, 7))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_test_pred,
    labels=model_classes,
    normalize="true",
    values_format=".2f",
    xticks_rotation=30,
    cmap=None,
    ax=ax
)

ax.set_title(
    "Multinomial Logistic Regression: "
    "Normalized Confusion Matrix"
)

plt.tight_layout()

plt.savefig(
    os.path.join(
        OUTPUT_FOLDER,
        "plot_multinomial_confusion_matrix_normalized.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.close()


# ============================================================
# 19. EXTRACT MULTINOMIAL COEFFICIENTS
# ============================================================

fitted_preprocessor = multinomial_model.named_steps[
    "preprocessor"
]

feature_names = fitted_preprocessor.get_feature_names_out()

# Make names easier to read
feature_names = [
    name.replace("numeric__", "")
        .replace("categorical__", "")
    for name in feature_names
]

coefficient_matrix = multinomial_model.named_steps[
    "model"
].coef_

intercepts = multinomial_model.named_steps[
    "model"
].intercept_

coefficient_rows = []

for class_index, class_name in enumerate(model_classes):
    for feature_index, feature_name in enumerate(feature_names):
        coefficient = coefficient_matrix[
            class_index,
            feature_index
        ]

        coefficient_rows.append({
            "Outcome_Class": class_name,
            "Predictor": feature_name,
            "Coefficient": coefficient,
            "Odds_Ratio": np.exp(coefficient),
            "Absolute_Coefficient": abs(coefficient)
        })

coefficient_results = pd.DataFrame(coefficient_rows)

coefficient_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_coefficients_and_odds_ratios.csv"
    ),
    index=False
)

intercept_results = pd.DataFrame({
    "Outcome_Class": model_classes,
    "Intercept": intercepts
})

intercept_results.to_csv(
    os.path.join(
        OUTPUT_FOLDER,
        "multinomial_intercepts.csv"
    ),
    index=False
)


# ============================================================
# 20. COEFFICIENT PLOTS
# ============================================================

# Create one plot for each behavior class
for class_name in model_classes:

    class_coefficients = (
        coefficient_results[
            coefficient_results["Outcome_Class"] == class_name
        ]
        .sort_values(
            "Absolute_Coefficient",
            ascending=False
        )
        .head(15)
        .sort_values("Coefficient")
    )

    plt.figure(figsize=(10, 7))

    plt.barh(
        class_coefficients["Predictor"],
        class_coefficients["Coefficient"]
    )

    plt.axvline(
        x=0,
        linewidth=1
    )

    plt.title(
        f"Top Multinomial Coefficients: {class_name}"
    )
    plt.xlabel("Standardized Logistic Regression Coefficient")
    plt.ylabel("Predictor")
    plt.tight_layout()

    safe_class_name = (
        str(class_name)
        .replace(" ", "_")
        .replace("/", "_")
    )

    plt.savefig(
        os.path.join(
            OUTPUT_FOLDER,
            f"plot_top_coefficients_{safe_class_name}.png"
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 21. MULTICLASS ROC CURVES
# ============================================================

try:
    y_test_binary = label_binarize(
        y_test,
        classes=model_classes
    )

    # label_binarize produces one column for a two-class problem.
    if len(model_classes) > 2:

        roc_rows = []

        plt.figure(figsize=(9, 7))

        for class_index, class_name in enumerate(model_classes):

            # A ROC curve requires both positive and negative examples.
            if len(np.unique(y_test_binary[:, class_index])) < 2:
                continue

            false_positive_rate, true_positive_rate, thresholds = roc_curve(
                y_test_binary[:, class_index],
                y_test_probability[:, class_index]
            )

            class_auc = auc(
                false_positive_rate,
                true_positive_rate
            )

            plt.plot(
                false_positive_rate,
                true_positive_rate,
                label=f"{class_name}: AUC = {class_auc:.3f}"
            )

            for fpr, tpr, threshold in zip(
                false_positive_rate,
                true_positive_rate,
                thresholds
            ):
                roc_rows.append({
                    "Behavior_Class": class_name,
                    "False_Positive_Rate": fpr,
                    "True_Positive_Rate": tpr,
                    "Threshold": threshold,
                    "AUC": class_auc
                })

        plt.plot(
            [0, 1],
            [0, 1],
            linestyle="--",
            label="Chance"
        )

        plt.title("One-vs-Rest Multiclass ROC Curves")
        plt.xlabel("False Positive Rate")
        plt.ylabel("True Positive Rate")
        plt.legend()
        plt.tight_layout()

        plt.savefig(
            os.path.join(
                OUTPUT_FOLDER,
                "plot_multinomial_roc_curves.png"
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        pd.DataFrame(roc_rows).to_csv(
            os.path.join(
                OUTPUT_FOLDER,
                "multinomial_roc_curve_values.csv"
            ),
            index=False
        )

        try:
            overall_auc = roc_auc_score(
                y_test_binary,
                y_test_probability,
                average="macro",
                multi_class="ovr"
            )

            pd.DataFrame({
                "Metric": ["Macro one-vs-rest ROC AUC"],
                "Value": [overall_auc]
            }).to_csv(
                os.path.join(
                    OUTPUT_FOLDER,
                    "multinomial_roc_auc.csv"
                ),
                index=False
            )

        except Exception as error:
            print("Overall ROC AUC could not be calculated:")
            print(error)

except Exception as error:
    print("\nROC curves could not be created:")
    print(error)


# ============================================================
# 22. OPTIONAL STATSMODELS MULTINOMIAL MODEL
# ============================================================
# This section attempts to provide traditional coefficients,
# standard errors, z statistics, and p-values.
#
# It may fail if:
# - the dataset is too small,
# - predictors are highly correlated,
# - there are too many predictors,
# - a category perfectly predicts the outcome.

try:
    # Transform the entire predictor dataset
    X_transformed = fitted_preprocessor.transform(X)

    X_statsmodels = pd.DataFrame(
        X_transformed,
        columns=feature_names,
        index=X.index
    )

    # Remove constant transformed columns
    nonconstant_columns = [
        column
        for column in X_statsmodels.columns
        if X_statsmodels[column].nunique() > 1
    ]

    X_statsmodels = X_statsmodels[
        nonconstant_columns
    ]

    # Add intercept
    X_statsmodels = sm.add_constant(
        X_statsmodels,
        has_constant="add"
    )

    # Convert target to numeric class codes
    y_statsmodels = pd.Categorical(
        y,
        categories=model_classes
    ).codes

    statsmodels_model = sm.MNLogit(
        y_statsmodels,
        X_statsmodels
    )

    statsmodels_result = statsmodels_model.fit(
        method="newton",
        maxiter=200,
        disp=False
    )

    # Save text summary
    with open(
        os.path.join(
            OUTPUT_FOLDER,
            "statsmodels_multinomial_summary.txt"
        ),
        "w"
    ) as file:
        file.write(statsmodels_result.summary().as_text())

    # Extract parameters
    params = statsmodels_result.params
    standard_errors = statsmodels_result.bse
    p_values = statsmodels_result.pvalues
    z_values = params / standard_errors

    statsmodels_rows = []

    # MNLogit uses the first category as the reference outcome.
    reference_class = model_classes[0]
    compared_classes = model_classes[1:]

    for equation_index, class_name in enumerate(compared_classes):

        for predictor_name in params.index:

            coefficient = params.iloc[
                params.index.get_loc(predictor_name),
                equation_index
            ]

            standard_error = standard_errors.iloc[
                standard_errors.index.get_loc(predictor_name),
                equation_index
            ]

            z_value = z_values.iloc[
                z_values.index.get_loc(predictor_name),
                equation_index
            ]

            p_value = p_values.iloc[
                p_values.index.get_loc(predictor_name),
                equation_index
            ]

            statsmodels_rows.append({
                "Reference_Outcome": reference_class,
                "Compared_Outcome": class_name,
                "Predictor": predictor_name,
                "Coefficient": coefficient,
                "Standard_Error": standard_error,
                "Z_Statistic": z_value,
                "P_Value": p_value,
                "Odds_Ratio": np.exp(coefficient)
            })

    statsmodels_results_df = pd.DataFrame(
        statsmodels_rows
    )

    statsmodels_results_df.to_csv(
        os.path.join(
            OUTPUT_FOLDER,
            "statsmodels_multinomial_coefficients_pvalues.csv"
        ),
        index=False
    )

    # Model-fit information
    statsmodels_fit = pd.DataFrame({
        "Measure": [
            "Log-likelihood",
            "Null log-likelihood",
            "Pseudo R-squared",
            "Likelihood-ratio statistic",
            "Likelihood-ratio p-value",
            "AIC",
            "BIC",
            "Number of observations"
        ],
        "Value": [
            statsmodels_result.llf,
            statsmodels_result.llnull,
            statsmodels_result.prsquared,
            statsmodels_result.llr,
            statsmodels_result.llr_pvalue,
            statsmodels_result.aic,
            statsmodels_result.bic,
            statsmodels_result.nobs
        ]
    })

    statsmodels_fit.to_csv(
        os.path.join(
            OUTPUT_FOLDER,
            "statsmodels_multinomial_fit_statistics.csv"
        ),
        index=False
    )

    print(
        "\nStatsmodels multinomial model was fitted successfully."
    )

except Exception as error:

    print(
        "\nStatsmodels multinomial model could not be fitted."
    )
    print(
        "The scikit-learn multinomial model was still completed."
    )
    print("Reason:")
    print(error)

    with open(
        os.path.join(
            OUTPUT_FOLDER,
            "statsmodels_model_error.txt"
        ),
        "w"
    ) as file:
        file.write(
            "The traditional statsmodels MNLogit model could not "
            "be fitted.\n\n"
        )
        file.write(
            "This commonly occurs because of perfect prediction, "
            "small category sizes, too many predictors, or strong "
            "multicollinearity.\n\n"
        )
        file.write(f"Error message:\n{str(error)}")


# ============================================================
# 23. CREATE AN EXCEL WORKBOOK
# ============================================================

excel_path = os.path.join(
    OUTPUT_FOLDER,
    "cat_behavior_analysis_results.xlsx"
)

with pd.ExcelWriter(
    excel_path,
    engine="openpyxl"
) as writer:

    data_summary.to_excel(
        writer,
        sheet_name="Data Summary",
        index=False
    )

    behavior_counts.to_excel(
        writer,
        sheet_name="Behavior Counts",
        index=False
    )

    combined_association_results.to_excel(
        writer,
        sheet_name="Individual Tests",
        index=False
    )

    sex_table.to_excel(
        writer,
        sheet_name="Sex Contingency"
    )

    sex_percent.to_excel(
        writer,
        sheet_name="Sex Percentages"
    )

    group_table.to_excel(
        writer,
        sheet_name="Group Contingency"
    )

    group_percent.to_excel(
        writer,
        sheet_name="Group Percentages"
    )

    age_descriptive.to_excel(
        writer,
        sheet_name="Age Descriptive",
        index=False
    )

    performance_results.to_excel(
        writer,
        sheet_name="Model Performance",
        index=False
    )

    coefficient_results.to_excel(
        writer,
        sheet_name="Model Coefficients",
        index=False
    )

    test_predictions.to_excel(
        writer,
        sheet_name="Test Predictions",
        index=False
    )


# ============================================================
# 24. WRITE A README FILE
# ============================================================

readme_text = f"""
CAT PROBLEMATIC-BEHAVIOR ANALYSIS
=================================

Target variable:
{TARGET_COLUMN}

Individual analyses:
1. {SEX_COLUMN} versus {TARGET_COLUMN}
   - Chi-square test
   - Bias-corrected Cramer's V
   - Contingency table
   - Stacked percentage plot

2. {AGE_COLUMN} versus {TARGET_COLUMN}
   - Spearman rank correlation
   - Kruskal-Wallis test
   - Descriptive statistics
   - Boxplot
   - Individual-value plot

3. {GROUP_COLUMN} versus {TARGET_COLUMN}
   - Chi-square test
   - Bias-corrected Cramer's V
   - Contingency table
   - Stacked percentage plot

Multivariable analysis:
- Multinomial logistic regression
- Every column except {TARGET_COLUMN} was considered as a predictor.
- Numerical missing values were replaced by the median.
- Categorical missing values were replaced by the most frequent category.
- Numerical predictors were standardized.
- Categorical predictors were dummy coded.
- The model used class-balanced logistic regression.
- A traditional statsmodels multinomial model was also attempted
  to obtain standard errors and p-values.

Behavior ordering:
{behavior_levels}

Important interpretation notes:
- Cramer's V measures association, not causation.
- Spearman correlation treats behavior levels as ordered.
- Multinomial logistic-regression coefficients compare outcome classes.
- Positive coefficients increase the estimated log odds of the listed
  outcome class, while negative coefficients decrease them.
- Odds ratios above 1 indicate increased odds.
- Odds ratios below 1 indicate decreased odds.
- Statistical significance does not establish causation.
"""

with open(
    os.path.join(
        OUTPUT_FOLDER,
        "README.txt"
    ),
    "w"
) as file:
    file.write(readme_text)


# ============================================================
# 25. CREATE ZIP FILE
# ============================================================

ZIP_PATH = "/content/cat_behavior_analysis_results.zip"

if os.path.exists(ZIP_PATH):
    os.remove(ZIP_PATH)

with zipfile.ZipFile(
    ZIP_PATH,
    "w",
    zipfile.ZIP_DEFLATED
) as zip_file:

    for root, directories, files in os.walk(OUTPUT_FOLDER):

        for filename in files:

            complete_path = os.path.join(
                root,
                filename
            )

            relative_path = os.path.relpath(
                complete_path,
                OUTPUT_FOLDER
            )

            zip_file.write(
                complete_path,
                arcname=relative_path
            )


# ============================================================
# 26. DISPLAY MAIN RESULTS
# ============================================================

print("\n" + "=" * 70)
print("INDIVIDUAL ASSOCIATION RESULTS")
print("=" * 70)

display(combined_association_results)

print("\n" + "=" * 70)
print("MULTINOMIAL MODEL PERFORMANCE")
print("=" * 70)

display(performance_results)

print("\n" + "=" * 70)
print("TOP MODEL COEFFICIENTS")
print("=" * 70)

display(
    coefficient_results
    .sort_values(
        ["Outcome_Class", "Absolute_Coefficient"],
        ascending=[True, False]
    )
    .groupby("Outcome_Class")
    .head(10)
)

print("\nAnalysis completed successfully.")
print("Results folder:", OUTPUT_FOLDER)
print("ZIP file:", ZIP_PATH)


# ============================================================
# 27. AUTOMATICALLY DOWNLOAD THE ZIP FILE
# ============================================================

from google.colab import files

files.download(ZIP_PATH)
```
